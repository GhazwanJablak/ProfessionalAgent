```python
from rich.console import Console
```

```python
def show(text):
    console = Console()
    try:
        console.print(text)
    except Exception:
        print(text)


def checklist_report(checklist, completed):
    result = ""
    for index, item in enumerate(checklist):
        if not completed[index]:
            result += f"Checklist #{index + 1}: {item}\n"
        else:
            result += f"Checklist #{index + 1}: [green][strike]{item}[/strike][/green]\n"
    show(result)
    return result


def create_checklist(description):
    checklist, completed = [], []
    checklist.extend(description)
    completed.extend([False] * len(description))
    return checklist_report(checklist, completed)


def mark_complete(completed, index, completion_notes):
    if 1 <= index <= len(completed):
        completed[index - 1]=True
    else:
        return "No item to complete"
    show(completion_notes)
    return completed
```

```python
checklist, completed = create_checklist(["Doing homework", "writing some code"])
result1 = checklist_report(checklist=checklist, completed=completed)
completed = mark_complete(completed=completed, index=1, completion_notes="Done")
result2 = checklist_report(checklist=checklist, completed=completed)
```
